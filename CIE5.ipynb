```python
import spacy
from nltk import ngrams

nlp = spacy.load("en_core_web_sm")

doc = nlp("AI is simulation of human intelligence by machine. It includes expert system NLP speech recognition and machine vision.")

bigrams = ngrams(doc, 2)
print(list(bigrams))

trigrams = ngrams(doc, 3)
print(list(trigrams))
```
